# Day 1: Pandas Basics (with NumPy comparison)

Topics: Series, DataFrame, inspecting data, reading and writing CSV, Pandas vs NumPy.

## 1. Import libraries

```python
import pandas as pd
import numpy as np
print(pd.__version__)
```

## 2. Series (1-D labelled data)

```python
s = pd.Series([85, 92, 78], index=["Asha", "Ravi", "Meena"])
print(s)
```

## 3. DataFrame (2-D table)

```python
data = {
    "name":  ["Asha", "Ravi", "Meena", "Kiran", "Divya", "Arjun"],
    "dept":  ["CSE", "ECE", "CSE", "ME", "ECE", "CSE"],
    "city":  ["Bengaluru", "Mysuru", "Bengaluru", "Hubli", "Mysuru", "Mangaluru"],
    "marks": [85, 92, 78, None, 88, 95],
    "age":   [20, 21, 20, 22, 21, 20],
}
df = pd.DataFrame(data)
print(df)
```

## 4. Inspect the data

```python
print(df.shape)       # (rows, columns)
print(df.dtypes)      # type of each column
df.info()             # columns, non-null counts, memory
```

```python
df.describe()          # count, mean, std, min, quartiles, max (numeric columns)
```

## 5. Write and read CSV

```python
df.to_csv("students.csv", index=False)       # write
back = pd.read_csv("students.csv")           # read
```

```python
back.head()
```

## 6. Pandas vs NumPy

NumPy arrays have no labels and one dtype. Pandas adds labels and mixed types.

```python
arr = np.array([[85, 20], [92, 21]])
print(arr)                              # no labels, one dtype
print(df[["marks", "age"]].head(2))     # labels, mixed types possible
```

```python
print(df["marks"].to_numpy())           # Pandas -> NumPy
print(np.mean(arr[:, 0]))                # NumPy maths on one column
```
