# NumPy Basics: Aggregations, Statistics, Correlation & Matrix Ops

```python
import numpy as np
from numpy.random import randint

np.random.seed(42)  # reproducible results
```

## 1. Sum, product and mean of a matrix

```python
mat1 = randint(1, 10, 9).reshape(3, 3)
mat2 = randint(1, 10, 9).reshape(3, 3)
print(mat1)
print(mat2)
```

```python
print(np.sum(mat1))   # sum of all elements
print(np.prod(mat1))  # product of all elements
```

```python
print(np.sum(mat1, axis=0))   # column-wise sum
print(np.sum(mat1, axis=1))   # row-wise sum
print(np.prod(mat1, axis=0))  # column-wise product
print(np.prod(mat1, axis=1))  # row-wise product
```

```python
print(np.mean(mat1))     # overall mean
print(np.mean(mat1[0]))  # mean of row 0
print(np.mean(mat1[1]))  # mean of row 1
```

## 2. Standard deviation, variance, median

```python
mat = randint(1, 100, 9).reshape(3, 3)
print(mat)
print("Std dev :", np.std(mat))
print("Variance:", np.var(mat))
```

```python
# np.median does not need a sorted 1D array, but we sort it to see the middle value
print(np.sort(mat.flatten()))
print("Median:", np.median(mat))
```

## 3. Correlation and scatter plots

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Two unrelated random arrays -> correlation close to 0
A = randint(1, 10, 20)
B = randint(1, 1000, 20)
print(np.corrcoef(A, B))

plt.scatter(A, B)
plt.title("Scatter plot of A vs. B")
plt.show()
```

```python
# B depends linearly on A (plus noise) -> correlation close to 1
A = randint(1, 50, 20)
B = 2 * A + 10 * np.random.randn(20)

plt.scatter(A, B)
plt.title("Scatter plot of A vs. B (linear relationship + noise)")
plt.show()

print(np.corrcoef(A, B))
```

```python
# A variable is perfectly correlated with itself
print(np.corrcoef(A, A))
```

## 4. Dot product and transpose

```python
A = randint(1, 10, 9).reshape(3, 3)
B = randint(1, 10, 9).reshape(3, 3)
print(A, "\n\n", B)
print("\nMatrix product:\n", np.dot(A, B))   # same as A @ B
```

```python
# Dot product of two 1D vectors gives a single number
A = randint(1, 10, 5)
B = randint(1, 10, 5)
print(A, "\n", B, "\n", np.dot(A, B))
```

```python
# Transpose of a 3x2 matrix -> 2x3
A = randint(1, 10, 6).reshape(3, 2)
print(A, "\n")
print(np.transpose(A))   # same as A.T
```
